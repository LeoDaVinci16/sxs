```python
import networkx as nx

G = nx.DiGraph()

edges = [
    ("8105", "0", 200),
    ("8107", "0", 200),
    ("8106", "0", 200),
    ("8108", "0", 200),
    ("8104", "0", 200),  # duplicate edge (kept as-is)
    ("0", "1", 1000),
    ("1", "1180", 100),
    ("1", "2", 900),
]

for u, v, value in edges:
    G.add_edge(u, v, flow=value)
```

```python
for n in G.nodes:
    inflow = sum(d["flow"] for u, v, d in G.in_edges(n, data=True))
    outflow = sum(d["flow"] for u, v, d in G.out_edges(n, data=True))

    print(n, "balance =", inflow - outflow)
```

```python
from pyvis.network import Network

net = Network(height="750px", width="100%", directed=True)

for n in G.nodes:
    net.add_node(n, label=str(n))

for u, v, d in G.edges(data=True):
    net.add_edge(
        u,
        v,
        value=d["flow"],
        title=f"flow: {d['flow']}"
    )

net.write_html("factory_network.html")
```

```python
import pandas as pd
from pyvis.network import Network

df = pd.read_csv("sankey_1.csv")

net = Network(height="750px", width="100%", directed=True)

# add nodes + edges
for _, row in df.iterrows():
    u = str(row["source"])
    v = str(row["target"])
    f = float(row["cabal"])

    # ensure nodes exist (PyVis auto-creates, but this allows styling later)
    net.add_node(u, label=u)
    net.add_node(v, label=v)

    net.add_edge(
        u,
        v,
        value=f,                 # thickness
        title=f"flow: {f}"      # hover tooltip
    )

net.write_html("factory_network.html", notebook=False)
```
