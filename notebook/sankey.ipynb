```python
from pyvis.network import Network

# -----------------------------
# 1. BUS DEFINITION (nodes)
# -----------------------------
# bus_id : (name, x, y)

buses = {
    "0":   ("Main",     0,   0),
    "1":   ("Splitter", 300, 0),
    "2":   ("Outlet A", 600, -100),
    "3":   ("Outlet B", 600, 100),
    "8105": ("Intake 1", -300, 150),
    "8107": ("Intake 2", -300, 50),
}

# -----------------------------
# 2. BRANCH DEFINITION (edges)
# -----------------------------
# (source, target, flow)

branches = [
    ("8105", "0", 200),
    ("8107", "0", 200),
    ("0", "1", 1000),
    ("1", "2", 400),
    ("1", "3", 600),
]

# -----------------------------
# 3. BUILD NETWORK
# -----------------------------
net = Network(height="750px", width="100%", directed=True)

# IMPORTANT: fixed layout (disable physics)
net.toggle_physics(False)

# -----------------------------
# 4. ADD NODES (BUSSES)
# -----------------------------
for bus_id, (name, x, y) in buses.items():
    net.add_node(
        bus_id,
        label=name,
        x=x,
        y=y,
        fixed=True
    )

# -----------------------------
# 5. ADD EDGES (BRANCHES)
# -----------------------------
max_flow = max(b[2] for b in branches)

for u, v, flow in branches:
    net.add_edge(
        u,
        v,
        value=flow,                 # thickness scaling
        title=f"flow: {flow}"
    )

# -------------------------------------------------
# 5. VISUAL OPTIONS (STRAIGHT + HIERARCHICAL)
# -------------------------------------------------
net.set_options("""
{
  "layout": {
    "hierarchical": {
      "enabled": true,
      "direction": "LR",
      "sortMethod": "directed",
      "levelSeparation": 180,
      "nodeSpacing": 120
    }
  },
  "physics": {
    "enabled": false
  },
  "edges": {
    "smooth": false
  }
}
""")

# -----------------------------
# 6. EXPORT
# -----------------------------
net.write_html("factory_network.html", notebook=False)
print("Saved: factory_network.html")
```
